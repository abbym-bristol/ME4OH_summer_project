# SHRED on ME4OH full-field OFAM3 SST data - Changing numbers of sensors

Experiment 2.5: changing number of sensors and comparing metrics

Note: Run training and evaluating halves of shred_ofam3sst_static_changing_sensor_nums.py script first

```python
import json

import matplotlib.pyplot as plt
import pandas as pd
```

```python
experiments = {"Anomaly SST": "v5-anomaly-gru", "Actual SST": "v5-gru"}

na_metrics = {}
gs_metrics = {}
for name, exp in experiments.items():
    with open(f"Reconstructions/change_num_sensors/{exp}/eval_metric_results.json", 'r') as f:
        data = json.load(f)

    # Unpack data dict into NA and GS dataframes
    na_metrics[name] = pd.DataFrame([data[key]['NA'] for key in sorted(data.keys(), key=lambda x: int(x))], 
                        index=[int(key) for key in sorted(data.keys(), key=lambda x: int(x))])
    gs_metrics[name] = pd.DataFrame([data[key]['GS'] for key in sorted(data.keys(), key=lambda x: int(x))], 
                      index=[int(key) for key in sorted(data.keys(), key=lambda x: int(x))])
    
    
```

```python
colours = {  # IBM colourblind palette
    "RMSE": "#1589e8",
    "MAE": "#dc267f",
    "R2": "#631ff3",
    "Average SSI": "#fe5100",
    "Average all time diff": "#ffb000"
}

line_style = {"Anomaly SST": "--", "Actual SST": ":"}
```

```python
for name, na_df in na_metrics.items():
    for col in na_df.columns:
        plt.plot(na_df.index, na_df[col], label=col+" "+name, c=colours[col], marker='.', ls=line_style[name])
plt.xlabel("Number of sensors")
plt.ylabel("Metric value")
plt.legend(bbox_to_anchor=(1,1))
```

```python
for col in ["RMSE", "MAE"]:
    for name, na_df in na_metrics.items():
        plt.plot(na_df.index, na_df[col], label=name, c=colours[col], marker='.', ls=line_style[name])
    plt.xlabel("Number of sensors")
    plt.ylabel(col)
    plt.legend(title="Trained with:")
    plt.show()
```

```python
for name, gs_df in gs_metrics.items():
    for col in ["RMSE", "MAE"]:
        plt.plot(gs_df.index, gs_df[col], label=col, c=colours[col], marker='.', ls=line_style[name])
plt.xlabel("Number of sensors")
plt.ylabel("Metric value")
plt.legend()
```
